```python
crude_oil_data = pd.read_csv('data/us_crude_oil_production.csv')
crude_oil_data.head(5)
```

```python
crude_oil_data.tail()
```

```python
crude_oil_data.shape
```

```python
crude_oil_data.columns
```

```python
crude_oil_data.columns[(crude_oil_data.sum(axis=0))==0]  #axis=0 is for rows, axis=1 for columns
```

```python
crude_oil_data.drop(['Arizona','Virginia'], inplace=True, axis=1) #inplace=True will not create a duplicate copy of the dataset and will make changes in the original data itself. Default inplace value is 'False'
```

```python
crude_oil_data.columns
```

```python
crude_oil_data['Date'] = pd.to_datetime(crude_oil_data['Month'])
crude_oil_data['Date'].head()
```

```python
crude_oil_data.drop(['Month'], inplace=True, axis=1)
crude_oil_data.columns
```

```python
crude_oil_data = crude_oil_data.rename(columns={'Federal Offshore Gulf of Mexico Crude Oil' : 'Mexico',
                                                'Federal Offshore Pacific Crude Oil':'Pacific' })
crude_oil_data.columns
```

```python
crude_oil_data['Year'] = crude_oil_data['Date'].dt.year
crude_oil_data['Month'] = crude_oil_data['Date'].dt.month
crude_oil_data['Year'].sample(10)
```

```python
crude_oil_data['Month'].sample(10)
```

```python
crude_oil_data.to_csv('crude_oil_data_processed.csv', index=False) #index=False will not create any index column in the csv file
```
